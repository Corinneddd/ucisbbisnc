| 数据类型 | 代表字段 | 实际含义 | 缺失值处理 | 常用转换 | 注意事项 |
|---|---|---|---|---|---|
| 标识型 | `Id` | 房屋编号 | 不处理 | 删除 | 不应作为普通特征 |
| 普通类别型 | `Neighborhood`、`HouseStyle`、`RoofStyle` | 不同类别，没有明确大小关系 | 填 `"None"` 或众数 | One-Hot Encoding | 不能直接转成 1、2、3 |
| 有序类别型 | `ExterQual`、`KitchenQual` | 有等级关系 | 无设施填 0 或 `"None"` | 映射为等级数字 | `Ex > Gd > TA > Fa > Po` |
| 二分类 | `CentralAir` | 是否有中央空调 | 填众数或 `"None"` | `Y/N → 1/0` | 可以直接转为 0/1 |
| 连续数值型 | `LotArea`、`GrLivArea` | 面积、金额等 | 中位数或 0 | `log1p`、标准化 | 面积通常右偏 |
| 离散计数型 | `BedroomAbvGr`、`Fireplaces` | 房间、车位、浴室数量 | 通常填 0 或中位数 | 保持整数或标准化 | 不要当作类别编码 |
| 年份型 | `YearBuilt`、`YrSold` | 建造、出售年份 | 中位数或业务规则 | 构造房龄 | 年份不一定直接使用 |
| 月份型 | `MoSold` | 销售月份 | 中位数或众数 | `sin/cos` 周期编码 | 12 月和 1 月是相邻月份 |
| 目标变量 | `SalePrice` | 房屋售价 | 通常不填充 | `log1p` | 预测后使用 `expm1` 恢复 |

# 处理缺失值

```python
# inspect missinh values
import pandas as pd
import numpy as np

df = pd.read_csv("data.csv")

print(df.shape)
print(df.info())
print(df.head())
```

```python
# how many missing values in each column
missing_count = df.isnull().sum()

missing_rate = df.isnull().mean() * 100

summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Rate (%)": missing_rate
})

print(summary.sort_values(
    "Missing Rate (%)",
    ascending=False
))
```
