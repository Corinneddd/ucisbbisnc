### Which Models Require Skewness Transformation?

| Model | Need to Handle Feature Skewness? | Reason |
|---|---|---|
| Linear Regression | Sometimes | May improve nonlinear relationships or reduce the influence of extreme values |
| Ridge / Lasso | Sometimes | May improve feature scaling and functional relationships |
| KNN | Worth considering | Skewed distributions and feature scales can affect distance calculations |
| Random Forest | Usually not necessary | Tree-based models split features based on thresholds |
| XGBoost | Usually not necessary | Tree-based models generally handle skewed features well |
| Neural Networks | Sometimes helpful | Transformations and scaling may improve optimization |

**Key Takeaway:** Skewness does not always require transformation. Whether to transform a feature depends on the model, the data distribution, and out-of-sample validation performance.
